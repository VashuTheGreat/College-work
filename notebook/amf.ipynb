```python

# utils
def save_to_drive(data, file_path):

    dir_name = os.path.dirname(file_path)

    if dir_name:
        os.makedirs(dir_name, exist_ok=True)

    if isinstance(data, torch.Tensor):
        data = data.detach().cpu().numpy()

    if isinstance(data, np.ndarray):
        data = pd.DataFrame(data)

    elif isinstance(data, pd.Series):
        data = data.to_frame()

    data.to_csv(file_path, index=False)
```

```python


@dataclass
class Config:

    alpha: float = 0.5
    beta: float = 0.5

    eps: float = 0.9
    min_sample: int = 3

    input_file_name: str = "./data/survey1.xls"

    output_folder_name: str = "./output"

    recon_err_file_path: str = os.path.join(
        output_folder_name,
        "pca_reconstruction_error.csv"
    )

    direc_score_file_path: str = os.path.join(
        output_folder_name,
        "pca_direction_score.csv"
    )

    ent_score_file_path: str = os.path.join(
        output_folder_name,
        "entropy_score.csv"
    )

    pca_score_file_path: str = os.path.join(
        output_folder_name,
        "pca_score.csv"
    )


    kMean_score_file_path: str = os.path.join(
        output_folder_name,
        "kmeans_score.csv"
    )

    


    dbscan_score_file_path: str = os.path.join(
        output_folder_name,
        "dbscan_score.csv"
    )

    

    
    rrf_k: int = 30

    pca_weight: float = 0.5
    kmean_weight: float = 0.3
    dbscan_weight: float = 0.1
    entrpy_weight:float=0.1
    hybrid_score_file_path: str = os.path.join(
        output_folder_name,
        "hybrid_rrf_score.csv"
    )

    def __post_init__(self):

        os.makedirs(
            self.output_folder_name,
            exist_ok=True
        )
```

```python


class Abstract(ABC):
    def __init__(self):
        super().__init__()
    def initiate(self):
        pass    
```

```python

class PCA_feature_extractor(Abstract):

    def __init__(self, config: Config):
        self.config = config

    
    @staticmethod
    def reconstruction_error(X, n_components=1):
        pca = PCA(n_components=n_components)

        X_np = X.detach().cpu().numpy()

        X_pca = pca.fit_transform(X_np)
        X_reconstructed = pca.inverse_transform(X_pca)

        X_reconstructed = torch.tensor(
            X_reconstructed,
            dtype=X.dtype,
            device=X.device
        )

        error = torch.mean((X - X_reconstructed) ** 2)

        return X_reconstructed, error

    @staticmethod
    def get_pc1(X, n_components=1):
        pca = PCA(n_components=n_components)

        X_np = X.detach().cpu().numpy()

        pca.fit(X_np)

        return torch.tensor(
            pca.components_[0],
            dtype=X.dtype,
            device=X.device
        )

    @staticmethod
    def leave_one_out_pca(X: torch.Tensor, beta=0.7):
        E_full = PCA_feature_extractor.reconstruction_error(X, n_components=1)[1]
        pc1_full = PCA_feature_extractor.get_pc1(X)

        n_features = X.shape[1]

        recon_scores = torch.zeros(n_features, device=X.device)
        direction_scores = torch.zeros(n_features, device=X.device)

        for i in range(n_features):
            mask = torch.arange(n_features, device=X.device) != i
            X_drop = X[:, mask]

            _, E_i = PCA_feature_extractor.reconstruction_error(X_drop, n_components=1)

            recon_scores[i] = E_i - E_full

            pc1_i = PCA_feature_extractor.get_pc1(X_drop)

            min_len = min(len(pc1_full), len(pc1_i))

            a = pc1_full[:min_len]
            b = pc1_i[:min_len]

            cos_sim = torch.dot(a, b) / (
                torch.norm(a) * torch.norm(b)
            )

            direction_scores[i] = 1 - cos_sim

        # final_scores = (
        #     beta * recon_scores +
        #     (1 - beta) * direction_scores
        # )

        return recon_scores, direction_scores


    # ========================== leave_one_out K-mean ============================================


    def initiate(self):

        recon_err, direc_score = PCA_feature_extractor.leave_one_out_pca(
            self.config.data,
            self.config.beta
        )

        # ent_score = PCA_feature_extractor.entropy_calculation(
        #     self.config.data
        # )

        save_to_drive(
            pd.DataFrame(
            zip(
                recon_err.cpu().numpy(),
                self.config.feature_names
            ),
            columns=["scores", "features"]
        ),
            self.config.recon_err_file_path
        )

        save_to_drive(
            pd.DataFrame(
            zip(
                direc_score.cpu().numpy(),
                self.config.feature_names
            ),
            columns=["scores", "features"]
        ),
            self.config.direc_score_file_path
        )

        # save_to_drive(
        #     pd.DataFrame(
        #     zip(
        #         ent_score.cpu().numpy(),
        #         self.config.feature_names
        #     ),
        #     columns=["scores", "features"]
        # ),
        #     self.config.ent_score_file_path
        # )

        pca_score = (
            self.config.beta * recon_err +
            (1 - self.config.beta) * direc_score
        )

        save_to_drive(
            pd.DataFrame(
            zip(
                pca_score.cpu().numpy(),
                self.config.feature_names
            ),
            columns=["scores", "features"]
        ),
            self.config.pca_score_file_path
        )

        
```

```python


class KMeans_feature_extractor(Abstract):

    def __init__(self, config: Config):
        self.config = config

    

    @staticmethod
    def find_best_k(X, k_range=(2, 10)):
        X_np = X.detach().cpu().numpy()

        scores = []

        for k in range(k_range[0], k_range[1] + 1):
            labels = KMeans(
                n_clusters=k,
                random_state=42,
                n_init=10
            ).fit_predict(X_np)

            score = silhouette_score(X_np, labels)

            scores.append(score)

        scores = torch.tensor(scores)

        best_k = torch.argmax(scores).item() + k_range[0]

        return best_k

    @staticmethod
    def feature_importance_kmeans(X):
        X_np = X.detach().cpu().numpy()

        best_k = KMeans_feature_extractor.find_best_k(X)

        labels_full = KMeans(
            n_clusters=best_k,
            random_state=42,
            n_init=10
        ).fit_predict(X_np)

        base_score = silhouette_score(X_np, labels_full)

        n_features = X.shape[1]

        masks = ~torch.eye(
            n_features,
            dtype=torch.bool,
            device=X.device
        )

        scores = torch.zeros(
            n_features,
            dtype=torch.float32,
            device=X.device
        )

        for i in range(n_features):
            X_drop = X[:, masks[i]]

            X_drop_np = X_drop.detach().cpu().numpy()

            labels = KMeans(
                n_clusters=best_k,
                random_state=42,
                n_init=10
            ).fit_predict(X_drop_np)

            score = silhouette_score(X_drop_np, labels)

            scores[i] = base_score - score

        return scores





    def initiate(self):

        score = KMeans_feature_extractor.feature_importance_kmeans(
            self.config.data
        )

        save_to_drive(
            pd.DataFrame(
            zip(
                score.cpu().numpy(),
                self.config.feature_names
            ),
            columns=["scores", "features"]
        ),
            self.config.kMean_score_file_path
        )

        # ent_score = KMeans_feature_extractor.entropy_calculation(
        #     self.config.data
        # )

        # save_to_drive(
        #     pd.DataFrame(
        #     zip(
        #         ent_score.cpu().numpy(),
        #         self.config.feature_names
        #     ),
        #     columns=["scores", "features"]
        # ),
        #     self.config.kMean_ent_file_path
        # )

        final_score = (score 
        )

        save_to_drive(
            pd.DataFrame(
            zip(
                final_score.cpu().numpy(),
                self.config.feature_names
            ),
            columns=["scores", "features"]
        ),
            self.config.kMean_score_file_path
        )
```

```python
class DBSCAN_feature_extractor(Abstract):

    def __init__(self, config: Config):
        self.config = config

    
    @staticmethod

    def feature_importance_dbscan(
        X,
        eps=0.5,
        min_sample=5
    ):
        X_np = X.detach().cpu().numpy()

        def safe_silhouette(data, labels):
            unique = set(labels) - {-1}

            if len(unique) < 2:
                return 0

            return silhouette_score(data, labels)

        labels_full = DBSCAN(
            eps=eps,
            min_samples=min_sample
        ).fit_predict(X_np)

        base_score = safe_silhouette(X_np, labels_full)

        n_features = X.shape[1]

        masks = ~torch.eye(
            n_features,
            dtype=torch.bool,
            device=X.device
        )

        scores = torch.zeros(
            n_features,
            dtype=torch.float32,
            device=X.device
        )

        for i in range(n_features):
            X_drop = X[:, masks[i]]

            X_drop_np = X_drop.detach().cpu().numpy()

            labels = DBSCAN(
                eps=eps,
                min_samples=min_sample
            ).fit_predict(X_drop_np)
            # print(np.unique(labels, return_counts=True))

            score = safe_silhouette(X_drop_np, labels)

            scores[i] = base_score - score

        return scores
    def initiate(self):

        score = DBSCAN_feature_extractor.feature_importance_dbscan(
            self.config.data,
            eps=self.config.eps,
            min_sample=self.config.min_sample
        )

        save_to_drive(
            pd.DataFrame(
            zip(
                score.cpu().numpy(),
                self.config.feature_names
            ),
            columns=["scores", "features"]
        ),
            self.config.dbscan_score_file_path
        )
```

```python

class Entropy_feature_extractor(Abstract):
    def __init__(self,config:Config):
        self.config=config

    def initiate(self):
        data=self.config.data
        entropies = []

        for i in range(data.shape[1]):
            col = data[:, i].detach().cpu().numpy()

            probs = pd.Series(col).value_counts(normalize=True)

            entropies.append(entropy(probs))

        final_score=torch.tensor(
            entropies,
            dtype=torch.float32,
            device=data.device
        )  
        save_to_drive(
            pd.DataFrame(
            zip(
                final_score.cpu().numpy(),
                self.config.feature_names
            ),
            columns=["scores", "features"]
        ),
            self.config.ent_score_file_path
        )    

        return final_score
```

```python

class PipelineRunner:

    def __init__(self, config: Config):

        self.config = config

        df = pd.read_csv(
            self.config.input_file_name,
        )
        df.dropna(inplace=True)

        df.drop_duplicates(
            subset='Name of organization',
            inplace=True
        )

        amf_cols = [
            col.split("AMF_ ")[1]
            for col in df.columns
            if col.startswith("AMF_")
        ]

        df.columns = [
            x if "_ " not in x
            else x.split("_ ")[1]
            for x in df.columns
        ]

        numeric_df = df.select_dtypes(include=['number'])

        sc = MinMaxScaler()

        scaled = sc.fit_transform(numeric_df)

        df[numeric_df.columns] = scaled
        df[numeric_df.columns] = np.square(
            df[numeric_df.columns]
        )

        # self.df = df

        self.data_cols = amf_cols
        self.config.feature_names=self.data_cols
        # self.arf_cols = arf_cols
        # self.apf_cols = apf_cols

        # self.data_cols = numeric_df.columns.tolist()

        self.data = torch.tensor(
            df[self.data_cols].values,
            dtype=torch.float32
        )

        # self.apf_data = torch.tensor(
        #     df[self.apf_cols].values,
        #     dtype=torch.float32
        # )

        # self.arf_data = torch.tensor(
        #     df[self.arf_cols].values,
        #     dtype=torch.float32
        # )

        # self.data = torch.tensor(
        #     df[self.data_cols].values,
        #     dtype=torch.float32
        # )

        self.config.data = self.data
        self.pca_pipe=PCA_feature_extractor(self.config)
        self.kmean_pipe = KMeans_feature_extractor(
            self.config
        )
        self.dbscan_pipe = DBSCAN_feature_extractor(
            self.config
        )

        self.entropy_pipe=Entropy_feature_extractor(self.config)
    def pca(self):

        self.pca_pipe.initiate()

    def k_mean(self):
      
        self.kmean_pipe.initiate()

    def dbscan(self):

        self.dbscan_pipe.initiate()
    
    def entrpy(self):

        self.entropy_pipe.initiate()

    def run_pipeline(self):
        print("input col sequences",self.data_cols)

        self.pca()

        self.k_mean()

        self.dbscan()
        self.entrpy()

    def hybrid(self):

        pca_df = pd.read_csv(
            self.config.pca_score_file_path
        )

        kmean_df = pd.read_csv(
            self.config.kMean_score_file_path
        )

        dbscan_df = pd.read_csv(
            self.config.dbscan_score_file_path
        )

        entrpy_df = pd.read_csv(
            self.config.ent_score_file_path
        )

        pca_df = pca_df.sort_values(
            by="scores",
            ascending=False
        ).reset_index(drop=True)

        kmean_df = kmean_df.sort_values(
            by="scores",
            ascending=False
        ).reset_index(drop=True)

        dbscan_df = dbscan_df.sort_values(
            by="scores",
            ascending=False
        ).reset_index(drop=True)

        entrpy_df = entrpy_df.sort_values(
            by="scores",
            ascending=False
        ).reset_index(drop=True)

        pca_df["rank"] = np.arange(1, len(pca_df) + 1)
        kmean_df["rank"] = np.arange(1, len(kmean_df) + 1)
        dbscan_df["rank"] = np.arange(1, len(dbscan_df) + 1)
        entrpy_df["rank"] = np.arange(1, len(entrpy_df) + 1)

        rrf_k = self.config.rrf_k

        pca_weight = self.config.pca_weight
        kmean_weight = self.config.kmean_weight
        dbscan_weight = self.config.dbscan_weight
        entrpy_weight = self.config.entrpy_weight

        pca_rrf = {
            row.features:
            pca_weight / (rrf_k + row.rank)

            for row in pca_df.itertuples()
        }

        kmean_rrf = {
            row.features:
            kmean_weight / (rrf_k + row.rank)

            for row in kmean_df.itertuples()
        }

        dbscan_rrf = {
            row.features:
            dbscan_weight / (rrf_k + row.rank)

            for row in dbscan_df.itertuples()
        }

        entrpy_rrf = {
            row.features:
            entrpy_weight / (rrf_k + row.rank)

            for row in entrpy_df.itertuples()
        }

        all_features = set(
            list(pca_rrf.keys()) +
            list(kmean_rrf.keys()) +
            list(dbscan_rrf.keys()) +
            list(entrpy_rrf.keys()) 
        )

        final_scores = []

        for feature in all_features:

            final_score = (
                pca_rrf.get(feature, 0) +
                kmean_rrf.get(feature, 0) +
                dbscan_rrf.get(feature, 0) +
                entrpy_rrf.get(feature, 0) 

            )

            final_scores.append(
                [feature, final_score]
            )

        final_df = pd.DataFrame(
            final_scores,
            columns=["features", "HybridScore"]
        )

        final_df = final_df.sort_values(
            by="HybridScore",
            ascending=False
        ).reset_index(drop=True)

        save_to_drive(
            final_df,
            self.config.hybrid_score_file_path
        )

        return final_df

    def print_everything(self):
        if not os.path.exists("output"):
            raise FileNotFoundError("No folder exist")

        print(os.listdir("output"))
        for file_name in os.listdir("output"):

            file_path = os.path.join(
                "output",
                file_name
            )

            df = pd.read_csv(file_path)
            

            print("\n")
            print("=" * 80)
            print(file_name)
            print("=" * 80)
            print(df)


```
