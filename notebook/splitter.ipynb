```python
import pandas as pd

data_file_path = "data/survey1.xls"


df = pd.read_excel(
    data_file_path,
    header=[1,2],
    index_col=0
)

ddf = df.copy()

df.columns = [
    f"{str(a).strip()}&{str(b).strip()}"
    for a,b in df.columns
]

df.columns = [
    col.replace(".1", "")
       .replace(".2", "")
       .replace(".3", "")
       .replace(".4", "")
    for col in df.columns
]

df.columns = [
    "1.1. Name of organization",
    "1.2. Type of business"
] + list(df.columns[2:])


df.columns = df.columns.map(
    lambda x:
        x.replace("1.1.", "")
         .replace("1.2.", "")
         .strip()

        if "&" not in x

        else f'{"".join(w[0] for w in x.split("&")[0].split())}_ {x.split("&")[1].strip()}'
)
df
```

```python
amf_col=list(df.columns[0:2])+[col for col in df.columns if "AMF" in col]
amf=df[amf_col]


arf_col=list(df.columns[0:2])+[col for col in df.columns if "ARF" in col]
arf=df[arf_col]


apf_col=list(df.columns[0:2])+[col for col in df.columns if "APF" in col]
apf=df[apf_col]

# =========== Saving to csv ===========

amf.to_csv("data/survey1_amf.csv", index=False)
arf.to_csv("data/survey1_arf.csv", index=False)
apf.to_csv("data/survey1_apf.csv", index=False)
```
